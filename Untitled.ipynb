```python
from pathlib import Path
import zipfile
import pandas as pd
import numpy as np

# Install if needed:
# %pip install -q pandas scikit-learn joblib

DATA_DIR = Path("datasets")
DATA_DIR.mkdir(exist_ok=True)

zip_candidates = list(Path(".").glob("DART_synthetic_training_data*.zip"))
if zip_candidates:
    with zipfile.ZipFile(zip_candidates[0], "r") as zf:
        zf.extractall(DATA_DIR)
    print("Extracted:", zip_candidates[0])

print("CSV files found:")
for p in DATA_DIR.rglob("*.csv"):
    print(" -", p)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score
import joblib
import json

AGENTS = {
    "grandparent": "grandparent_2500.csv",
    "elder_adult": "elder_adult_2500.csv",
    "youth": "youth_2500.csv",
    "children": "children_2500.csv",
}

TARGETS = [
    "task_time_seconds",
    "error_probability",
    "completion_probability",
    "difficulty_score",
]

# Keep these names synchronized with the DART Figma feature extractor.
FEATURES = [
    "font_size_px",
    "text_char_count",
    "node_width_px",
    "node_height_px",
    "contrast_ratio",
    "nearest_element_spacing_px",
    "visible_element_count",
    "navigation_depth",
    "label_complexity_score",
    "layout_complexity_score",
    "interactive_element_count",
    "text_node_count",
    "image_count",
    "minimum_target_dimension_px",
    "whitespace_ratio",
    "task_step_count",
    "reading_speed_wpm",
    "attention_span_sec",
    "visual_sensitivity",
    "motor_precision",
    "comprehension_factor",
]

MODEL_DIR = Path("models")
MODEL_DIR.mkdir(exist_ok=True)

metrics_rows = []
trained_models = {}

for agent_name, filename in AGENTS.items():
    csv_path = DATA_DIR / filename

    # Support the ZIP being extracted into a nested folder.
    if not csv_path.exists():
        matches = list(DATA_DIR.rglob(filename))
        if not matches:
            raise FileNotFoundError(
                f"Cannot find {filename}. Upload/extract DART_synthetic_training_data.zip "
                f"or place {filename} inside {DATA_DIR}."
            )
        csv_path = matches[0]

    df = pd.read_csv(csv_path)

    missing = [c for c in FEATURES + TARGETS if c not in df.columns]
    if missing:
        raise ValueError(f"{filename} is missing columns: {missing}")

    X = df[FEATURES].apply(pd.to_numeric, errors="coerce")
    y = df[TARGETS].apply(pd.to_numeric, errors="coerce")

    # Remove rows with missing values in required inputs/targets.
    valid = X.notna().all(axis=1) & y.notna().all(axis=1)
    X, y = X.loc[valid], y.loc[valid]

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.20, random_state=42
    )

    model = RandomForestRegressor(
        n_estimators=250,
        max_depth=None,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1,
    )
    model.fit(X_train, y_train)

    predictions = model.predict(X_test)

    # Store model and its expected input schema together.
    bundle = {
        "model": model,
        "agent_type": agent_name,
        "feature_columns": FEATURES,
        "target_columns": TARGETS,
        "training_rows": int(len(X_train)),
        "test_rows": int(len(X_test)),
        "data_type": "synthetic",
    }

    model_path = MODEL_DIR / f"{agent_name}_model.pkl"
    joblib.dump(bundle, model_path)
    trained_models[agent_name] = bundle

    for j, target in enumerate(TARGETS):
        metrics_rows.append({
            "agent": agent_name,
            "target": target,
            "MAE": mean_absolute_error(y_test.iloc[:, j], predictions[:, j]),
            "R2": r2_score(y_test.iloc[:, j], predictions[:, j]),
        })

    print(f"Saved {agent_name} model -> {model_path}")

metrics = pd.DataFrame(metrics_rows)
display(metrics.round(4))
print(done)
```
